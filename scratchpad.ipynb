```python
def save_to_disk(df:pd.DataFrame,title):
  base_dir = Path("./dataset") / title
  os.makedirs(base_dir,exist_ok=True)
  for index,row in df.iterrows():
    file_name = f"{index:05d}_{row['label']}.png"
    file_path = base_dir / file_name
    with open(file_path,'wb') as f:
      f.write(bytes(row['image']['bytes']))
  
```

```python
import torchvision.transforms.functional as F
```

```python
class DigitModel(nn.Module):
    def __init__(self, input_size=28 * 28, output=10):
        super().__init__()
        self.seq_layers = nn.Sequential(
            nn.Linear(input_size, 16),
            nn.ReLU(),
            nn.Linear(16, 16),
            nn.ReLU(),
            nn.Linear(16, output),
        )
        self.loss_function = nn.CrossEntropyLoss()
    def forward(self,x,labels=None):
      out = self.seq_layers(x)
      if labels is not None:
        loss = self.loss_function(out,labels)
      else:
        loss = None
        
      return out,loss
```

```python
def load_data(files = train_files):
  train_data = []
  for file in files:
    with Image.open(file) as img:
      label = file.stem.split("_")[1]
      label_tensor = torch.tensor([1 if i == int(label) else 0 for i in range(10)])
      train_data.append({'image':torch.tensor(np.array(img).reshape(28*28)),'label':label_tensor.float()})
  return train_data
```

```python
batch_size= 32
def get_batch(data=train_data):
  suffle_idx = torch.randperm(len(data))[:batch_size]
  # suffle_idx = range(batch_size)
  batch_dict= [data[i] for i in suffle_idx]
  image_tensor = torch.stack([item['image'].float()/255 for item in batch_dict]).to(device=device)
  label_tensor = torch.stack([item['label'] for item in batch_dict]).to(device=device)
  return image_tensor,label_tensor,batch_dict
  
```

```python
from PIL import Image,ImageTransform,ImageChops
import random
from PIL.ImageTransform import AffineTransform

def make_image_noisy(image,noise_level=40):
    """This function adds some noise to the image and rotate it randomly and translate it randomly
    Args:
        image (PIL.Image): The image to add noise to
        noise_level (float): The level of noise to add
    Returns:
        torch.Tensor: The noisy image
    """
    width, height = image.size
    angle = random.uniform(-10, 10)
    x_shift = random.uniform(-5, 5)
    y_shift = random.uniform(-5, 5)
    noise_texture = Image.effect_noise(image.size,noise_level)
    image = image.transform(image.size, AffineTransform([1, 0, x_shift, 0, 1, y_shift]))
    image = image.rotate(angle,resample=Image.Resampling.BICUBIC)
    image = image.resize((width, height),resample=Image.Resampling.BICUBIC) 
    image = ImageChops.multiply(image,noise_texture)
    return image
```
